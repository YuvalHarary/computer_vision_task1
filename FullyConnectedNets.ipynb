## Inline Question 1:
Did you notice anything about the comparative difficulty of training the three-layer network vs. training the five-layer network? In particular, based on your experience, which network seemed more sensitive to the initialization scale? Why do you think that is the case?

## Answer:
Yes: the five-layer network was much harder to train and much more sensitive to the initialization scale. The three-layer network reached 100% training accuracy for a fairly wide range of weight scales and learning rates, while the five-layer network trained only in a narrow band of weight scales (around 5e-2). With a smaller scale, such as the initial 1e-5, the loss stayed at about 2.3 (chance) and nothing was learned, and with a larger scale the loss exploded. The reason is that every layer multiplies the activations in the forward pass, and the gradients in the backward pass, by roughly weight_scale times the square root of the layer width. With more layers these factors compound exponentially: a scale slightly too small makes the signal and gradients vanish, and a scale slightly too large makes them explode. The shallower network has fewer such factors, so it tolerates a much wider range of initialization scales.

## Inline Question 2:

AdaGrad, like Adam, is a per-parameter optimization method that uses the following update rule:

```
cache += dw**2
w += - learning_rate * dw / (np.sqrt(cache) + eps)
```

John notices that when he was training a network with AdaGrad that the updates became very small, and that his network was learning slowly. Using your knowledge of the AdaGrad update rule, why do you think the updates would become very small? Would Adam have the same issue?


## Answer:
AdaGrad's `cache` is a running sum of all squared gradients seen so far, so it only grows and never decreases. The effective learning rate of each weight, `learning_rate / (sqrt(cache) + eps)`, therefore keeps shrinking during training, and after many steps the updates become very small and learning almost stops, even if the network is still far from a good solution. Adam does not have this problem: instead of a sum, it keeps an exponentially decaying average of the squared gradients (`v = beta2 * v + (1 - beta2) * dw**2`). Old gradients are gradually forgotten, so `v` follows the recent size of the gradients instead of growing forever, and the step size stays roughly `learning_rate`. Its bias correction also makes the early steps the right size.
